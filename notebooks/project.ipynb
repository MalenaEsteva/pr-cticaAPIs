# Práctica de API request

```python
import requests

usuario = 'torvalds'


def obtener_repositorios(usuario):
    """Consulta la API de GitHub y devuelve la lista de repositorios del usuario."""
    url = f'https://api.github.com/users/{usuario}/repos'
    respuesta = requests.get(url)
    respuesta.raise_for_status()
    return respuesta.json()


datos = obtener_repositorios(usuario)
```

## Exploración de la respuesta

Antes de procesar los datos, revisamos el tipo, la cantidad de repositorios
obtenidos y la estructura de un elemento de ejemplo.

```python
print(type(datos))
print(len(datos))
print(datos[0])
```

## Tarea 2: lista de repositorios

Se genera una lista de diccionarios con las claves `name`, `language`, `stars` y `forks`,
utilizando comprensión de listas.

```python
repositorios = [{'name': repo['name'],
                  'language': repo['language'],
                  'stars': repo['stargazers_count'],
                  'forks': repo['forks_count']}
                 for repo in datos]

print(repositorios)
```

## Tarea 3.1: repositorio más popular

Se busca el repositorio con mayor cantidad de estrellas.

```python
repo_popular = max(repositorios, key=lambda repo: repo['stars'])

print(f"Most popular repository: {repo_popular['name']}")
print(f"Stars: {repo_popular['stars']}")
```

## Tarea 3.2: lenguajes utilizados

Se obtienen los diferentes lenguajes de programación presentes en los repositorios,
sin repetir.

```python
lenguajes = sorted({repo['language'] for repo in repositorios if repo['language']})

print('Languages:')
for lenguaje in lenguajes:
    print(f'- {lenguaje}')
```

## Tarea 3.3: cantidad de repositorios por lenguaje

Se cuenta cuántos repositorios utilizan cada lenguaje, recorriendo la lista
con un ciclo `while`.

```python
conteo_lenguajes = {}
indice = 0

while indice < len(repositorios):
    lenguaje = repositorios[indice]['language']
    if lenguaje:
        conteo_lenguajes[lenguaje] = conteo_lenguajes.get(lenguaje, 0) + 1
    indice += 1

for lenguaje, cantidad in conteo_lenguajes.items():
    print(f'{lenguaje}: {cantidad}')
```
